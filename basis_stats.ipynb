```python
import pandas as pd
from scipy import stats

# Load dataset (replace with your dataset path or URL)
file_path = r"C:\Users\PC\Downloads\financials.xls"
df = pd.read_csv(file_path)

# Select only numerical columns
numeric_df = df.select_dtypes(include=['int64', 'float64'])

# Calculate statistics
mean_values = numeric_df.mean()
median_values = numeric_df.median()
mode_values = numeric_df.mode().iloc[0]   # mode() can return multiple rows
std_values = numeric_df.std()

# Display results
print("Mean:\n", mean_values)
print("\nMedian:\n", median_values)
print("\nMode:\n", mode_values)
print("\nStandard Deviation:\n", std_values)
```

```python
correlation = df['Price'].corr(df['Market Cap'])
print("Correlation between Price and Market Cap:", correlation)

correlation2 = df['Dividend Yield'].corr(df['Price/Earnings'])
print("Correlation between Dividend Yield and Price/Earnings:", correlation2)
```
